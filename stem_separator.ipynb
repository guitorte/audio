```python
import torch
import os
import glob

# --- Passo 1: Detecção de Hardware ---
if torch.cuda.is_available():
  device = "cuda"
  print("✅ GPU encontrada! Usando 'cuda' para processamento rápido.")
else:
  device = "cpu"
  print("⚠️ Nenhuma GPU encontrada. O processo será executado na CPU (Lento).")

# --- Passo 2: Encontrar Arquivos de Áudio ---
# Certifique-se de que a variável 'input_folder' foi definida anteriormente no seu código
search_pattern = os.path.join(input_folder, '*.mp3')
track_paths = glob.glob(search_pattern)

if not track_paths:
  print(f"\nERRO: Nenhum arquivo .mp3 foi encontrado em '{input_folder}'.")
else:
  print(f"\nEncontrado(s) {len(track_paths)} arquivo(s) .mp3 para processar com 6 STEMS:")
  for path in track_paths:
    print(f"- {os.path.basename(path)}")

  files_to_process = " ".join([f'"{path}"' for path in track_paths])

  # --- Passo 3: Executar o Comando Demucs (6 STEMS) ---
  print("\nIniciando a separação em 6 faixas (Vocais, Bateria, Baixo, Guitarra, Piano, Outros)...")

  # ALTERAÇÕES FEITAS AQUI:
  # 1. Modelo alterado para: -n htdemucs_6s
  # 2. Removido: --two-stems=vocals
  !python3 -m demucs -d "{device}" -n htdemucs_6s --mp3 --mp3-bitrate 320 -o "{output_folder}" {files_to_process}

  print("\n✅ Separação em 6 faixas concluída!")
```
